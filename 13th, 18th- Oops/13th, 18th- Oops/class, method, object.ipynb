```python
class computer():        
    
    def __init__(self): 
        print('in init')
    
    def config(self):     
        print('i5', '8gb')

com1 = computer('i5', 8) 
com2 = computer('i7', 16) 

com1.config() 
com2.config() 
```

```python
class computer:        
    
    def __init__(self,cpu,ram): # cpu & ram as argument
        print('in init')
    
    def config(self):     
        print('i5', '8gb')
        print('i7', 16)

com1 = computer('i5', 8)
com2 = computer('i7', 16)

com1.config() 
com2.config()
```

```python
class computer:        
    
    def __init__(self,cpu,ram, price): # cpu & ram as argument
        print('in init')
    
    def config(self):     
        print('i5', '8gb', 70000)
        print('i7', 16, 80000)

com1 = computer('i5', 8, 70000)
com2 = computer('i7', 16, 80000)

com1.config() 
com2.config()
```

```python
class computer():       
    
    def __init__(self,cpu,ram): # cpu & ram as argument
        self.cpu = cpu #hear self is the object & we need to refer to the object
        self.ram = ram 

  
    def config(self):     
        print('config is', self.cpu, self.ram)

com1 = computer('i5', 8)
com2 = computer('i7', 16)


com1.config() 
com2.config()
```

```python
class computer():       
    
    def __init__(self,cpu,ram,price,memory): # cpu & ram as argument
        self.cpu = cpu #hear self is the object & we need to refer to the object
        self.ram = ram
        self.price = price 
        self.memory = memory

  
    def config(self):     
        print('config is', self.cpu, self.ram, self.price, self.memory)

com1 = computer('i5', 8, 40, 256)
com2 = computer('i7', 16, 50, 1)


com1.config() 
com2.config()
```

```python
class computer():
    def __init__(self):
        self.name = 'akash'
        self.age = 37

c1 = computer()
c2 = computer() 

print(c1.name, c1.age)
print(c2.name, c2.age)
```
