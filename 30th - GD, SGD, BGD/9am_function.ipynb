```python
a = 20  

def something():
    #global a
    a = 15     
    print('in function',a)
    a = 9 

something()

print('out function',a)
```

```python
a = 20  

def something():
    global a
    a = 15     
    print('in function',a)
    a = 9 

something()

print('out function',a)
```

```python
x = 10  # Global variable

def update_x():
    #global x  # Declare that we are using the global variable x
    x += 5    # Modify the global variable

update_x()
print(x)  # Output: 15
```

```python
x = 10  # Global variable

def update_x():
    global x  # Declare that we are using the global variable x
    x += 5    # Modify the global variable

update_x()
print(x)  # Output: 15
```

```python
x = 10  # Global variable

def update_x():
    globals()['x'] += 6  # Access and modify the global variable using the dictionary

update_x()
print(x)  # Output: 15
```
